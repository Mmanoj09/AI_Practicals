```python
initial_state = {"At_A"}

goal = "At_C"

actions = [
    {
        "name": "Move A to B",
        "pre": {"At_A"},
        "add": {"At_B"},
        "del": {"At_A"}
    },
    {
        "name": "Move B to C",
        "pre": {"At_B"},
        "add": {"At_C"},
        "del": {"At_B"}
    }
]

def find_action(goal_condition):
    # Search for an action that adds the goal condition
    for action in actions:
        if goal_condition in action["add"]:
            return action
    return None

def make_plan(goal, state, plan):
    # Goal is already true
    if goal in state:
        return True

    # Find an action that can achieve the goal
    action = find_action(goal)

    if action is None:
        return False

    # First achieve the action's preconditions
    for condition in action["pre"]:
        if not make_plan(condition, state, plan):
            return False

    # Perform the action
    state.difference_update(action["del"])
    state.update(action["add"])
    plan.append(action["name"])

    return True

state = initial_state.copy()
plan = []

make_plan(goal, state, plane)
print("plan:")
for step in plan:
    print(step)



transitions = {
    'A': {'right': 'B', 'down': 'C'},
    'B': {'left': 'A', 'down': 'D'},
    'C': {'left': 'B', 'down': 'E'},
    'G':{}


}
rewards = {
    'A': -1,
    'C': -1,
    'B': -1,
    'E': 10
}



values={
    'A':0,
    'B':0,
    'C':0,
    'E':0
}

print(values)



gamma = 0.9  # Define the discount factor

for i in range(5):
    new_values = values.copy()

    for state in ['A', 'B', 'C']:
        possible_values = []

        for action, next_state in transitions[state].items():
            # Use .get() to prevent KeyError if a state is missing from rewards or values
            reward = rewards.get(next_state, 0)
            next_val = values.get(next_state, 0)
            value = reward + gamma * next_val
            possible_values.append(value)

        if possible_values:
            new_values[state] = max(possible_values)

    values = new_values

print("Final values:")
print(values)
```
